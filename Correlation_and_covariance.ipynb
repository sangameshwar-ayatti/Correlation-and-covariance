#Q1. Define Covariance and explain how it differs from Correlation in terms of scale and interpretation.

Ans -

Covariance measures how two variables change together.

If both variables increase or decrease together → positive covariance
If one increases while the other decreases → negative covariance
If there is no consistent pattern → covariance is near zero

#Q2. What does a positive, negative, and zero covariance indicate about the relationship between two variables.

Ans -

Positive Covariance (> 0)
Both variables move in the same direction
When one increases, the other also increases (and vice versa)

Example:

Study hours ↑ → Marks ↑

Indicates a direct relationship

Negative Covariance (< 0)
Variables move in opposite directions
When one increases, the other decreases

Example:

Price ↑ → Demand ↓

Indicates an inverse relationship

Zero Covariance (≈ 0)
No linear relationship between variables
Changes in one variable do not predict changes in the other

Example:

Shoe size and exam marks

Indicates no association (linear)

#Q3. Discuss the limitations of covariance as a measure of relationship between two variables. Why is correlation preferred in many cases.

Ans -

🔹 Limitations of Covariance

Covariance is useful, but it has several important drawbacks:

1. Depends on scale (units problem)

Covariance values change if the units change (e.g., meters → centimeters).
This makes it hard to interpret or compare across datasets.

2. No fixed range

Covariance can take any value (positive or negative, large or small).
There’s no standard scale to judge whether a relationship is strong or weak.

3. Difficult to interpret magnitude

A value like 200 or 0.5 doesn’t clearly tell how strong the relationship is.
It only gives direction, not a clear sense of strength.

4. Not suitable for comparison

You cannot compare covariance between different variable pairs because values depend on units and scale.

#Q4. Explain the difference between Pearson’s correlation coefficient and Spearman’s rank correlation coefficient. When would you prefer to use Spearman’s correlation

Ans -

🔹 Pearson’s Correlation Coefficient (r)
Measures the linear relationship between two variables
Uses actual numerical values
Assumes data is:
Normally distributed
Linearly related
Sensitive to outliers

 Example: Relationship between height and weight

🔹 Spearman’s Rank Correlation Coefficient (ρ or rs)
Measures the monotonic relationship (increasing or decreasing, not necessarily linear)
Uses ranks instead of actual values
Does not require normal distribution
Less affected by outliers

Example: Ranking students based on marks in two subjects

#Q5. If the correlation coefficient between two variables X and Y is 0.85, interpret this value in context. Can you infer causation from this value? Why or why not?

Ans -

🔹 Interpretation of r = 0.85

A correlation coefficient of 0.85 indicates a:

Strong positive relationship between X and Y
As X increases, Y tends to increase as well
The relationship is close to linear, though not perfectly (since it’s not exactly 1)

In practical terms: the variables move together in a consistent and strong way

🔹 Why not?
Third (lurking) variable
A hidden factor may influence both X and Y
Example: Ice cream sales and drowning incidents both increase due to temperature
Direction problem

Even if related, you don’t know:
Does X cause Y?
Or does Y cause X?

Coincidence
Sometimes variables appear related just by chance, especially in small datasets
